```python
df_clean["arrival_date"] = pd.to_datetime(
    df_clean["arrival_date_day_of_month"].astype(str)
    + "-"
    + df_clean["arrival_date_month"]
    + "-"
    + df_clean["arrival_date_year"].astype(str),
    format="%d-%B-%Y",
    errors="coerce"
)

print(
    "Invalid arrival dates:",
    df_clean["arrival_date"].isna().sum()
)

display(
    df_clean[
        [
            "arrival_date_year",
            "arrival_date_month",
            "arrival_date_day_of_month",
            "arrival_date"
        ]
    ].head(10)
)
```

```python
daily_bookings = (
    df_clean.groupby("arrival_date")
    .size()
    .reset_index(name="booking_count")
    .sort_values("arrival_date")
)

plt.figure(figsize=(16, 6))

plt.plot(
    daily_bookings["arrival_date"],
    daily_bookings["booking_count"]
)

plt.title("Daily Hotel Booking Arrivals")
plt.xlabel("Arrival Date")
plt.ylabel("Number of Reservations")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
monthly_bookings = (
    df_clean.groupby(
        df_clean["arrival_date"].dt.to_period("M")
    )
    .size()
)

plt.figure(figsize=(15, 6))

monthly_bookings.plot(kind="line", marker="o")

plt.title("Monthly Hotel Booking Arrivals")
plt.xlabel("Month")
plt.ylabel("Number of Reservations")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
room_comparison = pd.crosstab(
    df_clean["reserved_room_type"],
    df_clean["assigned_room_type"]
)

display(room_comparison)
```

```python
completed_stays = df_clean[
    df_clean["reservation_status"] == "Check-Out"
].copy()

completed_stays["stay_nights"] = (
    completed_stays["stays_in_weekend_nights"]
    + completed_stays["stays_in_week_nights"]
)

completed_stays = completed_stays[
    completed_stays["stay_nights"] > 0
].copy()

print("Completed stays:", len(completed_stays))

display(
    completed_stays[
        [
            "hotel",
            "arrival_date",
            "stay_nights",
            "assigned_room_type"
        ]
    ].head()
)
```

```python
expanded = completed_stays[
    [
        "hotel",
        "arrival_date",
        "stay_nights",
        "assigned_room_type"
    ]
].copy()

expanded["occupied_date"] = expanded.apply(
    lambda row: pd.date_range(
        start=row["arrival_date"],
        periods=int(row["stay_nights"]),
        freq="D"
    ).tolist(),
    axis=1
)

expanded = expanded.explode("occupied_date")

expanded = expanded.rename(
    columns={
        "assigned_room_type": "room_category"
    }
)

display(expanded.head(10))

```

```python
daily_demand = (
    expanded.groupby(
        [
            "occupied_date",
            "hotel",
            "room_category"
        ]
    )
    .size()
    .reset_index(name="occupied_room_nights")
)

daily_demand = daily_demand.sort_values(
    ["occupied_date", "hotel", "room_category"]
)

display(daily_demand.head(20))

print("Daily demand records:", len(daily_demand))
```

```python
daily_demand["occupied_date"] = pd.to_datetime(
    daily_demand["occupied_date"]
)

all_dates = pd.date_range(
    start=daily_demand["occupied_date"].min(),
    end=daily_demand["occupied_date"].max(),
    freq="D"
)

hotel_categories = daily_demand[
    ["hotel", "room_category"]
].drop_duplicates()

calendar = pd.DataFrame({
    "occupied_date": all_dates
})

calendar["_key"] = 1
hotel_categories["_key"] = 1

full_grid = calendar.merge(
    hotel_categories,
    on="_key"
).drop(columns="_key")

daily_demand = full_grid.merge(
    daily_demand,
    on=["occupied_date", "hotel", "room_category"],
    how="left"
)

daily_demand["occupied_room_nights"] = (
    daily_demand["occupied_room_nights"].fillna(0)
)

daily_demand = daily_demand.sort_values(
    ["hotel", "room_category", "occupied_date"]
)

display(daily_demand.head(20))
```

```python
OUTPUT_PATH = Path(
    "../data/processed/daily_room_demand.csv"
)

OUTPUT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

daily_demand.to_csv(
    OUTPUT_PATH,
    index=False
)

print("Processed dataset saved:", OUTPUT_PATH)
```
